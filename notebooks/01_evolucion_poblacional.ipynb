## Actividad 3 — Estadística descriptiva

Usamos solo los datos históricos (`Annual change in population`). Se toma el año 2020 y se calculan estadísticos sobre todas las entidades ese año.

```python
anio_seleccionado = 2020
df_anio = df[df["Year"] == anio_seleccionado][["Entity", "Year", "Annual change in population"]]
df_anio.head()
```

```python
cambio_poblacion = df_anio["Annual change in population"]

media = cambio_poblacion.mean()
mediana = cambio_poblacion.median()
varianza = cambio_poblacion.var()
desviacion_estandar = cambio_poblacion.std()
minimo = cambio_poblacion.min()
maximo = cambio_poblacion.max()

print(f"Media: {media:,.2f}")
print(f"Mediana: {mediana:,.2f}")
print(f"Varianza: {varianza:,.2f}")
print(f"Desviación estándar: {desviacion_estandar:,.2f}")
print(f"Mínimo: {minimo:,.2f}")
print(f"Máximo: {maximo:,.2f}")
```

```python
# chequeo rápido con numpy
print(np.mean(cambio_poblacion), np.std(cambio_poblacion, ddof=1))
```

```python
# Entidades con el valor mínimo y máximo en 2020
entidad_minimo = df_anio.loc[df_anio["Annual change in population"].idxmin(), "Entity"]
entidad_maximo = df_anio.loc[df_anio["Annual change in population"].idxmax(), "Entity"]

print("Entidad con el valor mínimo:", entidad_minimo)
print("Entidad con el valor máximo:", entidad_maximo)
```

### Preguntas

1. ¿Media? ~2,124,909.
2. ¿Mediana? ~34,293 (mucho más baja que la media, por los agregados grandes como "World").
3. ¿Desviación estándar? ~9,522,990, bastante alta por la diferencia de tamaño entre países/agregados.
4. ¿Mínimo y máximo? Mínimo: Venezuela (caída fuerte por la crisis migratoria). Máximo: World (es el agregado global).
